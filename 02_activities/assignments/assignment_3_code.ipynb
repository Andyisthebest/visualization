```python
    import numpy as np 
    import matplotlib.pyplot as plt 
    import pandas as pd 
    import scipy
    import PIL 
    import requests
```

```python
years = [2022, 2023, 2024, 2025, 2026]
fatals = [20, 15, 17, 18, 23]

automobile = [6, 4, 4, 7, 3]
cyclist = [0, 1, 4, 0, 0]
motorcycle = [1, 2, 0, 2, 6]
pedestrian = [13, 8, 9, 9, 14]

age_groups = ["< 17", "18 to 24", "25 to 34",
              "35 to 44", "45 to 54", "55 to 64", "65 +"]
age_counts = [1, 1, 4, 7, 1, 2, 7]

# -----------------------------
# Dashboard
# -----------------------------

fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# -----------------------------
# 1. Bar chart
# -----------------------------

axs[0, 0].bar(years, fatals)
axs[0, 0].set_title("Total Fatals by Year")
axs[0, 0].set_xlabel("Year")
axs[0, 0].set_ylabel("Fatal Count")

for x, y in zip(years, fatals):
    axs[0, 0].text(x, y + 0.3, str(y), ha="center")

# -----------------------------
# 2. Line chart
# -----------------------------

axs[0, 1].plot(years, automobile, marker="o", label="Automobile")
axs[0, 1].plot(years, cyclist, marker="o", label="Cyclist")
axs[0, 1].plot(years, motorcycle, marker="o", label="Motorcycle")
axs[0, 1].plot(years, pedestrian, marker="o", label="Pedestrian")

axs[0, 1].set_title("Total Fatals by Year and Involvement")
axs[0, 1].set_xlabel("Year")
axs[0, 1].set_ylabel("Fatal Count")
axs[0, 1].legend()

# -----------------------------
# 3. Pie chart
# -----------------------------

totals = [
    sum(automobile),
    sum(cyclist),
    sum(motorcycle),
    sum(pedestrian)
]

labels = ["Automobile", "Cyclist", "Motorcycle", "Pedestrian"]

axs[1, 0].pie(
    totals,
    labels=labels,
    autopct="%1.1f%%"
)

axs[1, 0].set_title("Total Fatals by Involvement")

# -----------------------------
# 4. Age group bar chart
# -----------------------------

axs[1, 1].bar(age_groups, age_counts)

axs[1, 1].set_title("Total Fatals by Age Group")
axs[1, 1].set_xlabel("Age Group")
axs[1, 1].set_ylabel("Fatal Count")

for i, value in enumerate(age_counts):
    axs[1, 1].text(i, value + 0.2, str(value), ha="center")

plt.tight_layout()
plt.show()
```
