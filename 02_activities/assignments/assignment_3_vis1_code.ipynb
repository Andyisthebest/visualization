```python
    import numpy as np 
    import matplotlib.pyplot as plt 
    import pandas as pd 
    import scipy
    import PIL 
    import requests
```

```python
# -----------------------------
# Data
# -----------------------------

years = [2022, 2023, 2024, 2025, 2026]
fatals = [20, 15, 17, 18, 23]

automobile = [6, 4, 4, 7, 3]
cyclist = [0, 1, 4, 0, 0]
motorcycle = [1, 2, 0, 2, 6]
pedestrian = [13, 8, 9, 9, 14]

age_groups = ["< 17", "18 to 24", "25 to 34",
              "35 to 44", "45 to 54", "55 to 64", "65 +"]
age_counts = [1, 1, 4, 7, 1, 2, 7]

# -----------------------------
# Dashboard (2 plots only)
# -----------------------------

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# -----------------------------
# 1. Bar chart
# -----------------------------

axs[0].bar(years, fatals)
axs[0].set_title("Total Fatals by Year")
axs[0].set_xlabel("Year")
axs[0].set_ylabel("Fatal Count")

for x, y in zip(years, fatals):
    axs[0].text(x, y + 0.3, str(y), ha="center")

# -----------------------------
# 2. Line chart
# -----------------------------

axs[1].plot(years, automobile, marker="o", label="Automobile")
axs[1].plot(years, cyclist, marker="o", label="Cyclist")
axs[1].plot(years, motorcycle, marker="o", label="Motorcycle")
axs[1].plot(years, pedestrian, marker="o", label="Pedestrian")

axs[1].set_title("Total Fatals by Year and Involvement")
axs[1].set_xlabel("Year")
axs[1].set_ylabel("Fatal Count")
axs[1].legend()

# -----------------------------
# Layout
# -----------------------------

plt.tight_layout()
plt.show()
```
